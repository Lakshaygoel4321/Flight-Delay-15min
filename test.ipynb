```python
import pickle
import numpy as np
import pandas as pd

with open('ohe_dest.pkl','rb') as file:
    ohe_dest = pickle.load(file)

with open('ohe_origin.pkl','rb') as file:
    ohe_origin = pickle.load(file)

with open('ohe_unique.pkl','rb') as file:
    ohe_unique = pickle.load(file)

with open('scaler.pkl','rb') as file:
    scaler = pickle.load(file)

with open('knn.pkl','rb') as file:
    knn = pickle.load(file)
```

```python
Month = int(input('Provide the Month'))

DayofMonth = int(input('Provide the Date of Month'))

DepTime = int(input("Provide the DepTime"))
```

```python
input_data = {
    'Month':Month,
    'DayofMoenth':DayofMonth,
    'DepTime':DepTime,
    'Origin':'SLC',	
    'Dest':'SEA',
    'Distance':689
}

```

```python
origin_df = ohe_origin.transform([[input_data['Origin']]]).toarray()
origin_pd = pd.DataFrame(origin_df,columns=ohe_origin.get_feature_names_out())
origin_pd
```

```python
dest_df = ohe_dest.transform([[input_data['Dest']]]).toarray()
dest_pd = pd.DataFrame(dest_df,columns=ohe_dest.get_feature_names_out())
dest_pd
```
